```python
# 파이썬 시각화 패키지 불러오기
import matplotlib.pyplot as plt
%matplotlib inline


# 사용자 운영체제 확인
import platform
platform.system()

# 운영체제별 한글 폰트 설정
if platform.system() == 'Darwin': # Mac 환경 폰트 설정
    plt.rc('font', family='AppleGothic')
elif platform.system() == 'Windows': # Windows 환경 폰트 설정
    plt.rc('font', family='Malgun Gothic')

plt.rc('axes', unicode_minus=False) # 마이너스 폰트 설정

import csv
import matplotlib.pyplot as plt
car = open("car_accidents.csv")
data = csv.reader(car)

result = []
month = ["1월", "2월", "3월", "4월", "5월", "6월", "7월", "8월", "9월", "10월", "11월", "12월"]
jan = 0
feb = 0
mar = 0
apr = 0
may = 0
jun = 0
jul = 0
aug = 0
ocb = 0
sep = 0
nov = 0
dec = 0

jan_death = 0
feb_death = 0
mar_death = 0
apr_death = 0
may_death = 0
jun_death = 0
jul_death = 0
aug_death = 0
ocb_death = 0
sep_death = 0
nov_death = 0
dec_death = 0
#가해자연령층, 발생월, 사고건수, 사망자수, 중상자수, 경상자수, 부상신고자수

#월별 교통사고 발생시 사망확률

for row in data:
    a = int(row[1])
    if a >= 12:
        for i in row[2:3]:
            dec = dec + int(i)
        for i in row[3:4]:
            dec_death = dec_death + int(i)
    elif a >=11:
        for k in row[2:3]:
            nov = nov + int(k)
        for k in row[3:4]:
            nov_death = nov_death + int(k)
    elif a >=10:
        for l in row[2:3]:
            sep = sep + int(l)
        for l in row[3:4]:
            sep_death = sep_death + int(l)
    elif a >=9:
        for m in row[2:3]:
            ocb = ocb + int(m)
        for m in row[3:4]:
            ocb_death = ocb_death + int(m)
    elif a >=8:
        for n in row[2:3]:
            aug = aug + int(n)
        for n in row[3:4]:
            aug_death = aug_death + int(n)
    elif a >=7:
        for o in row[2:3]:
            jul = feb + int(o)
        for o in row[3:4]:
            jul_death = jul_death + int(o)
    elif a >=6:
        for p in row[2:3]:
            jun = jun + int(p)
        for p in row[3:4]:
            jun_death = jun_death + int(p)
    elif a >=5:
        for q in row[2:3]:
            may = may + int(q)
        for q in row[3:4]:
            may_death = may_death + int(q)
    elif a >=4:
        for r in row[2:3]:
            apr = apr + int(r)
        for r in row[3:4]:
            apr_death = apr_death + int(r)
    elif a >=3:
        for s in row[2:3]:
            mar = mar + int(s)
        for s in row[3:4]:
            mar_death = mar_death + int(s)
    elif a >=2:
        for t in row[2:3]:
            feb = feb + int(t)
        for t in row[3:4]:
            feb_death = feb_death + int(t)
    elif a >=1:
        for u in row[2:3]:
            jan = jan + int(u)
        for z in row[3:4]:
            jan_death = jan_death + int(z)

result.append(jan_death/jan*100)
result.append(feb_death/feb*100)
result.append(mar_death/mar*100)
result.append(apr_death/apr*100)
result.append(may_death/may*100)
result.append(jun_death/jun*100)
result.append(jul_death/jul*100)
result.append(aug_death/aug*100)
result.append(ocb_death/ocb*100)
result.append(sep_death/sep*100)
result.append(nov_death/nov*100)
result.append(dec_death/dec*100)
        
        
plt.plot(month, result, color = "purple")
plt.title("2020년 교통사고 발생 시 사망 확률")
plt.xlabel("월")
plt.ylabel("사망할 확률")
plt.show()
```
