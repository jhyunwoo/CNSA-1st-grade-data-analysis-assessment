```python
# 파이썬 시각화 패키지 불러오기
import matplotlib.pyplot as plt
%matplotlib inline


# 사용자 운영체제 확인
import platform
platform.system()

# 운영체제별 한글 폰트 설정
if platform.system() == 'Darwin': # Mac 환경 폰트 설정
    plt.rc('font', family='AppleGothic')
elif platform.system() == 'Windows': # Windows 환경 폰트 설정
    plt.rc('font', family='Malgun Gothic')

plt.rc('axes', unicode_minus=False) # 마이너스 폰트 설정

import csv
import numpy as np
import matplotlib.pyplot as plt
crime = open("대검찰청_범죄발생지_20181231.csv")
data = csv.reader(crime)

#93
#도시별 범죄 발생 건수

result = []
type_of_crime = []

for row in data:
    for i in row[0:1]:
        if i not in type_of_crime:
            type_of_crime.append(str(i))
    for j in row[93:94]:
        result.append(int(j))
x = np.arange(len(type_of_crime))
plt.figure(figsize=(50,50), dpi=400)
plt.barh(type_of_crime, result,color = "red")
plt.title("천안시 범죄 유형별 발생 건수")
plt.xlabel("발생건수")
plt.ylabel("범죄 종류")
plt.show()
```
